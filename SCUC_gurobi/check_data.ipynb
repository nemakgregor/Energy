```python
import json
import pandas as pd
from pyomo.environ import *
```

```python
data_path = "C:/Users/egor1/Desktop/Energy/SCUC/data"
data_name = "case118-initcond.json"
```

```python
full_data_path = f"{data_path}/{data_name}"
with open(full_data_path, "r") as f:
    data = json.load(f)
```

```python
list(data.keys())
```

```python
parameters = data["Parameters"]
generators = pd.DataFrame(data["Generators"]).T
lines = pd.DataFrame(data["Transmission lines"]).T
contingencies = pd.DataFrame(data["Contingencies"]).T
buses = pd.DataFrame(data["Buses"]).T
reserves = pd.DataFrame(data["Reserves"]).T
```

```python
print(f"- Parameters: {parameters}")
print(f"- Generators: {generators.shape[0]} units")
print(f"- Transmission Lines: {lines.shape[0]}")
print(f"- Contingencies: {contingencies.shape[0]}")
print(f"- Buses: {buses.shape[0]}")
print(f"- Reserve Requirements: {reserves.to_dict(orient='records')}")
```
